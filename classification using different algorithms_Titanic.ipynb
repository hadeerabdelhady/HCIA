to handle missing values :
-> for coloumns which has so much missing values we will drop this coloumn
->for numerical coloumns that has little missing values we will fill the rest by median because it is robust to outliers and represent the middle age 
->for categorical coloumns that has little missing values we will fill these entries by mode the most repeatable value

```python
train_df["Embarked"].mode()#return series 
```

```python
train_df["Embarked"].mode()[0]
```

```python
train_df.drop(columns=['Cabin'], inplace=True)
test_df.drop(columns=['Cabin'], inplace=True)
train_df.drop(columns=["Ticket"],inplace=True)
test_df.drop(columns=["Ticket"],inplace=True)
train_df.drop(columns=["Name"],inplace=True)
test_df.drop(columns=["Name"],inplace=True)
train_df["Age"]=train_df["Age"].fillna(train_df["Age"].median())
test_df["Age"]=test_df["Age"].fillna(test_df["Age"].median())
train_df["Embarked"]=train_df["Embarked"].fillna(train_df["Embarked"].mode()[0])
test_df["Fare"]=test_df["Fare"].fillna(test_df["Fare"].median())
```

```python
encoder=LabelEncoder()
train_df["Sex"]=encoder.fit_transform(train_df["Sex"])
test_df["Sex"]=encoder.transform(test_df["Sex"])
```

pd.get_dummies() is a function in pandas that turns categorical values (like text) into numbers — specifically into separate columns that only contain 0s and 1s.
This is called one-hot encoding.

```python
train_df=pd.get_dummies(train_df,columns=["Embarked"],drop_first=True)
test_df=pd.get_dummies(test_df,columns=["Embarked"],drop_first=True)
```

```python
train_df[["Embarked_Q","Embarked_S"]]=train_df[["Embarked_Q","Embarked_S"]].astype(int)
test_df[["Embarked_Q","Embarked_S"]]=test_df[["Embarked_Q","Embarked_S"]].astype(int)
```

```python
train_df.head()
```

```python
test_df.head()
```

```python
scalar= MinMaxScaler()
train_df['Age']=scalar.fit_transform(train_df[["Age"]])
test_df["Age"]=scalar.transform(test_df[["Age"]])
```

```python
train_df['Fare']=scalar.fit_transform(train_df[["Fare"]])
test_df["Fare"]=scalar.transform(test_df[["Fare"]])
```

```python
x=train_df.drop(columns=["Survived","PassengerId"])
y=train_df["Survived"]

```

```python
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
y_train
```

```python
regressor1=LogisticRegression(max_iter=100)
regressor1.fit(x_train,y_train)
y_pred1=regressor1.predict(x_test)
```

```python
accuracy1=accuracy_score(y_test,y_pred1)
precision1=precision_score(y_test,y_pred1)
recall1=recall_score(y_test,y_pred1)
print(f"the accuracy is {accuracy1}")
print(f"the precision is {precision1}")
print(f"the recall is {recall1}")
```

```python
regressor2=LogisticRegression(max_iter=500)
regressor2.fit(x_train,y_train)
y_pred2=regressor2.predict(x_test)
```

```python
accuracy2=accuracy_score(y_test,y_pred2)
precision2=precision_score(y_test,y_pred2)
recall2=recall_score(y_test,y_pred2)
print(f"the accuracy is {accuracy2}")
print(f"the precision is {precision2}")
print(f"the recall is {recall2}")
```

```python
tree=DecisionTreeClassifier()
tree.fit(x_train,y_train)
y_pred3=tree.predict(x_test)
```

```python
accuracy3=accuracy_score(y_test,y_pred3)
precision3=precision_score(y_test,y_pred3)
recall3=recall_score(y_test,y_pred3)
print(f"the accuracy is {accuracy3}")
print(f"the precision is {precision3}")
print(f"the recall is {recall3}")
```

```python
forest=RandomForestClassifier(n_estimators=500)
forest.fit(x_train,y_train)
y_pred4=forest.predict(x_test)
```

```python
accuracy4=accuracy_score(y_test,y_pred4)
precision4=precision_score(y_test,y_pred4)
recall4=recall_score(y_test,y_pred4)
print(f"the accuracy is {accuracy4}")
print(f"the precision is {precision4}")
print(f"the recall is {recall4}")
```

```python
forest2=RandomForestClassifier(n_estimators=1000)
forest2.fit(x_train,y_train)
y_pred5=forest2.predict(x_test)
```

```python
accuracy5=accuracy_score(y_test,y_pred5)
precision5=precision_score(y_test,y_pred5)
recall5=recall_score(y_test,y_pred5)
print(f"the accuracy is {accuracy5}")
print(f"the precision is {precision5}")
print(f"the recall is {recall5}")
```

```python
forest3=RandomForestClassifier(n_estimators=3000)
forest3.fit(x_train,y_train)
y_pred6=forest3.predict(x_test)
```

```python
accuracy6=accuracy_score(y_test,y_pred6)
precision6=precision_score(y_test,y_pred6)
recall6=recall_score(y_test,y_pred6)
print(f"the accuracy is {accuracy6}")
print(f"the precision is {precision6}")
print(f"the recall is {recall6}")
```

```python


comparison_df = pd.DataFrame({
    '   Model': ['Logistic Regression with 100 iterations',"logistic regression with 500 iterations", 'Decision Tree', 'Random Forest with 500',"Random Forest with 1000","Random Forest with 3000"],
    '   Accuracy': [accuracy1, accuracy2, accuracy3,accuracy4,accuracy5,accuracy6],
    '   Precision': [precision1, precision2, precision3,precision4,precision5,precision6],
    '   Recall': [recall1, recall2, recall3,recall4,recall5,recall6],
})

print(comparison_df)
```
